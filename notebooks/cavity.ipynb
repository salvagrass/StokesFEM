# Lid Driven Cavity


We consider the stationary Navier-Stokes problem defined in the unit square domain $\Omega := {[0,1]}^2$
$$
\begin{cases}
  -2\mu\boldsymbol{\nabla \cdot} \boldsymbol{(\nabla_s u)} + \boldsymbol{\nabla} p = 0,
  \qquad&\text{in }\Omega, \\
  \boldsymbol{\nabla} \cdot \boldsymbol{u} = 0,
  \qquad &\text{in }\Omega, \\
  \boldsymbol{u} = 1\boldsymbol{i} + 0\boldsymbol{j}, \qquad&\text{on }\Gamma^{\text{up}} = \{ 0 \le x \le 1, \, y=1 \}, \\
  \boldsymbol{u} = \boldsymbol{0}, \qquad&\text{on }\partial\Omega \backslash \Gamma^{\text{up}}.
\end{cases}
$$
where $\mu = 1.0$, $\boldsymbol{\nabla_s u} := \boldsymbol{\frac{1}{2}(\nabla u + (\nabla u)^\top)}$

```python
from problems.cavity import generate_problem
from solvers import SolverFactory
from common import MethodType
from fenics import Constant,plot
from postprocessing import plot_combined
```

```python
problem = generate_problem(n=15)
plot(problem.mesh)
```

```python
solver = SolverFactory.create_solver(MethodType.MINI)
solution = solver.solve(problem, Constant(1.0))

fig, axes = plot_combined(solution)
```

```python
solver = SolverFactory.create_solver(MethodType.KS)
solution = solver.solve(problem, Constant(1.0))

fig, axes = plot_combined(solution)
```

```python
solver = SolverFactory.create_solver(MethodType.AUG)
solution = solver.solve(problem,Constant(1.0))

plot_combined(solution,'streamlines',layout='horizontal')
```
